```python
# ============================================================
# NAÏVE BAYES CLASSIFICATION
# Confusion Matrix and Accuracy
# Google Colab - No CSV Required
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)


# ------------------------------------------------------------
# 1. Load Iris Dataset
# ------------------------------------------------------------

iris = load_iris()

X = iris.data
y = iris.target

print("IRIS DATASET")
print("=" * 60)

print("Number of samples :", X.shape[0])
print("Number of features:", X.shape[1])

print("\nFeature Names:")
print(iris.feature_names)

print("\nTarget Classes:")
print(iris.target_names)


# ------------------------------------------------------------
# 2. Split Dataset
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# ------------------------------------------------------------
# 3. Create Naïve Bayes Model
# ------------------------------------------------------------

model = GaussianNB()


# ------------------------------------------------------------
# 4. Train the Model
# ------------------------------------------------------------

model.fit(X_train, y_train)

print("\nNaïve Bayes model trained successfully!")


# ------------------------------------------------------------
# 5. Make Predictions
# ------------------------------------------------------------

y_pred = model.predict(X_test)


# ------------------------------------------------------------
# 6. Calculate Accuracy
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("MODEL ACCURACY")
print("=" * 60)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)


# ------------------------------------------------------------
# 7. Generate Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


# ------------------------------------------------------------
# 8. Display Confusion Matrix Graphically
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Naïve Bayes Confusion Matrix")

plt.show()


# ------------------------------------------------------------
# 9. Classification Report
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=iris.target_names
    )
)


# ------------------------------------------------------------
# 10. Predict a New Sample
# ------------------------------------------------------------

# [Sepal Length, Sepal Width, Petal Length, Petal Width]

new_sample = np.array([
    [5.1, 3.5, 1.4, 0.2]
])

prediction = model.predict(
    new_sample
)

predicted_class = iris.target_names[
    prediction[0]
]

print("\n" + "=" * 60)
print("NEW SAMPLE PREDICTION")
print("=" * 60)

print("New Sample:", new_sample[0])

print(
    "Predicted Class:",
    predicted_class
)
```
