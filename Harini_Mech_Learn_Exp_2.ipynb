```python
# ============================================================
# CANDIDATE ELIMINATION ALGORITHM
# No CSV upload required
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. Create Training Dataset
# ------------------------------------------------------------

data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes']
],
columns=[
    'Sky',
    'AirTemp',
    'Humidity',
    'Wind',
    'Water',
    'Forecast',
    'EnjoySport'
])

print("TRAINING DATA")
print("=" * 60)
print(data.to_string(index=False))


# ------------------------------------------------------------
# 2. Separate Attributes and Target
# ------------------------------------------------------------

attributes = list(data.columns[:-1])

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values

n = len(attributes)

print("\nAttributes:", attributes)
print("Target:", data.columns[-1])


# ------------------------------------------------------------
# 3. Candidate-Elimination Algorithm
# ------------------------------------------------------------

# Specific boundary
S = ['0'] * n

# General boundary
G = [['?'] * n]


# ------------------------------------------------------------
# Function: Check whether hypothesis covers example
# ------------------------------------------------------------

def covers(hypothesis, example):

    for h, x in zip(hypothesis, example):

        if h == '0':
            return False

        if h != '?' and h != x:
            return False

    return True


# ------------------------------------------------------------
# Function: Check whether h1 is more general than h2
# ------------------------------------------------------------

def more_general_or_equal(h1, h2):

    for a, b in zip(h1, h2):

        if a != '?' and a != b:
            return False

    return True


# ------------------------------------------------------------
# Function to remove duplicate hypotheses
# ------------------------------------------------------------

def remove_duplicates(hypotheses):

    result = []

    for h in hypotheses:

        if h not in result:
            result.append(h)

    return result


# ------------------------------------------------------------
# 4. Process Training Examples
# ------------------------------------------------------------

for index in range(len(X)):

    example = X[index]
    label = str(y[index]).strip().lower()

    print("\n" + "=" * 60)
    print("TRAINING EXAMPLE", index + 1)
    print("=" * 60)

    print("Example :", list(example))
    print("Class   :", y[index])


    # ========================================================
    # POSITIVE EXAMPLE
    # ========================================================

    if label == 'yes':

        # Remove G hypotheses that don't cover example
        G = [
            g for g in G
            if covers(g, example)
        ]

        # Generalize S
        for i in range(n):

            if S[i] == '0':
                S[i] = example[i]

            elif S[i] != example[i]:
                S[i] = '?'


    # ========================================================
    # NEGATIVE EXAMPLE
    # ========================================================

    else:

        new_G = []

        for g in G:

            if covers(g, example):

                # Specialize G
                for i in range(n):

                    if g[i] == '?':

                        # Use value from S
                        if S[i] != '0' and S[i] != '?':

                            new_h = g.copy()

                            new_h[i] = S[i]

                            # New hypothesis must not cover
                            # negative example
                            if not covers(new_h, example):

                                new_G.append(new_h)

            else:

                new_G.append(g)

        G = remove_duplicates(new_G)


    # --------------------------------------------------------
    # Display current boundaries
    # --------------------------------------------------------

    print("\nSpecific Boundary (S):")
    print(S)

    print("\nGeneral Boundary (G):")

    if len(G) == 0:
        print("EMPTY")
    else:
        for g in G:
            print(g)


# ------------------------------------------------------------
# 5. Final Result
# ------------------------------------------------------------

print("\n\n" + "=" * 60)
print("FINAL VERSION SPACE")
print("=" * 60)

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")

for g in G:
    print(g)

print("\nCandidate-Elimination Algorithm completed successfully.")
```
